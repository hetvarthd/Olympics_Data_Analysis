```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
players = ["Chris Gayle", "Kieron Pollard", "Dwayne Bravo"]
chris = [292, 608, 733, 708, 196, 491]
polly = [273, 146, 220, 420, 273, 419]
bravo = [61, 11, 371, 121, 8, 195]
years = np.array([i for i in range(2010, 2016)])
total_runs = [4965, 3412, 1560]
width = 0.25
```

```python
plt.bar(years - width, chris, width = width, label = players[0])
plt.bar(years, polly, width = width, label = players[1])
plt.bar(years + width, bravo, width = width, label = players[2])
plt.legend()
plt.xlabel("IPL Years")
plt.ylabel("Total Runs")
plt.title("Performance Comparison from 2010 to 2015")
```

```python
bars = plt.barh(players, total_runs, color='purple')
plt.xlabel("Total Runs")
plt.title("IPL Career Runs")
plt.tight_layout()
for bar in bars:
    width = bar.get_width()
    plt.text(width+50, bar.get_y() + bar.get_height()/2, f'{width}', va='center')
```
