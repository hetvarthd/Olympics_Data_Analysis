```python
athletes_df = athletes.merge(regions, how='left', on='NOC')
athletes_df.head()
```

# 4. Standardizing Column Names

# 14. Female Participation Trends Across Olympic Games

```python
female_summer = athletes_df[(athletes_df['Sex']=='F')&(athletes_df['Season']=='Summer')].filter(['Sex', 'Year'], axis=1)
female_summer = female_summer.groupby('Year').count().reset_index()
female_summer
```

```python
female_winter = athletes_df[(athletes_df['Sex']=='F')&(athletes_df['Season']=='Winter')].filter(['Sex', 'Year'], axis=1)
female_winter = female_winter.groupby('Year').count().reset_index()
female_winter
```

```python
plt.figure(figsize=(12,6))

plt.plot(female_winter['Year'], female_winter['Sex'], linewidth = 2, color='#239FA9', label = 'Winter Olympics')
plt.plot(female_summer['Year'], female_summer['Sex'], linewidth = 2, color='#F06956', label = 'Summer Olympics')
plt.title('Female Participation Trends Across Olympic Games')
plt.xlabel('Olympics Event Year')
plt.ylabel('Total No.of Female Participants')
plt.legend()
plt.grid(True)

ax = plt.gca()
ax.set_facecolor('#FBE7B8')
```

# 16. Rio Olympics 2016 Performance Analysis

```python
rio_olympics = athletes_df[athletes_df['City']=='Rio de Janeiro']
rio_olympics
```

### 16.1 Top 5 Countries with Most Medals

```python
rio_top_5_countries = rio_olympics[rio_olympics['Medal'].notnull()]['Region'].value_counts().head(5)
rio_top_5_countries
```

### 16.2 No.of Gold Medals of Countries with Most Medals

```python
rio_gold = rio_olympics[rio_olympics['Medal']=='Gold']['Region'].value_counts().reset_index(name='Medal')
rio_gold = rio_gold[rio_gold['Region'].isin(rio_top_5_countries.index)]
```

### 16.3 No.of Silver Medals of Countries with Most Medals

```python
rio_silver = rio_olympics[rio_olympics['Medal']=='Silver']['Region'].value_counts().reset_index(name='Medal')
rio_silver = rio_silver[rio_silver['Region'].isin(rio_top_5_countries.index)]
```

### 16.4 No.of Bronze Medals of Countries with Most Medals

```python
rio_bronze = rio_olympics[rio_olympics['Medal']=='Bronze']['Region'].value_counts().reset_index(name='Medal')
rio_bronze = rio_bronze[rio_bronze['Region'].isin(rio_top_5_countries.index)]
```

### 16.5 Plotting Medal Distribution by Country

```python
ind = np.arange(5)  # the x locations for the groups
width = 0.25        # the width of the bars

fig = plt.figure(figsize=(12,6))
ax = fig.add_subplot(111)

rects1 = ax.bar(ind, rio_gold['Medal'].tolist(), width, color='#D4AF37')
rects2 = ax.bar(ind+width, rio_silver['Medal'].tolist(), width, color='#C0C0C0')
rects3 = ax.bar(ind+(2*width), rio_bronze['Medal'].tolist(), width, color='#CD7F32')

ax.set_ylabel('No.of Medals')
ax.set_xticks(ind+width)
ax.set_xticklabels(tuple(rio_top_5_countries.index.tolist()))
ax.legend((rects1[0], rects2[0], rects3[0]), ('Gold', 'Silver', 'Bronze'))
ax.set_facecolor('#F5F5F5')

def autolabel(rects):
    for rect in rects:
        h = rect.get_height()
        ax.text(rect.get_x()+rect.get_width()/2, h, '%d'%int(h), ha='center', va='bottom')

autolabel(rects1)
autolabel(rects2)
autolabel(rects3)

plt.show()
```
