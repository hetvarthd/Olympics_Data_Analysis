```python
import matplotlib.pyplot as plt
```

```python
mob_OS = ["Android", "iOS", "Others"]
shares = [71.55, 26.93, 1.52]
colors = ['#ff9999','#66b3ff','#99ff99']
explode = [0, 0.1, 0]
```

```python
plt.style.use('ggplot')
plt.title("Mobile OS Market Share (Early 2026)")
plt.pie(shares, 
        labels=mob_OS, 
        colors=colors, 
        wedgeprops={'edgecolor': 'black', 'linewidth': 1, 'linestyle': '-'},
        explode=explode,
        shadow=True,
        startangle=190,
        autopct='%1.2f%%')
```
