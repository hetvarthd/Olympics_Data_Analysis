```python
import matplotlib.pyplot as plt
```

### All T20 CWC Event Years:

```python
years = [2007, 2009, 2010, 2012, 2014, 2016, 2021, 2022, 2024, 2026]
```

### Scores of Rohit, Virat & Hardik:

```python
rohit = [88, 131, 84, 82, 200, 88, 174, 116, 257]
virat = [185, 319, 273, 68, 296, 151]
hardik = [16, 69, 128, 144, 217]
```

### Graph Plotting:

```python
plt.style.use('petroff10')
plt.plot(years[:-1], rohit, 'o-', linewidth = 2, label = 'Rohit Sharma')
plt.plot(years[3:-1], virat, 's-', linewidth = 2, label = 'Virat Kohli')
plt.plot(years[-5:], hardik, 'D-', linewidth = 2, label = 'Hardik Pandya')
plt.title('T20 CWC Runs Comparison')
plt.xlabel('T20 CWC Event Year')
plt.ylabel('Total Runs')
plt.xticks(years)
plt.legend()
plt.grid(True)
plt.tight_layout()
```
