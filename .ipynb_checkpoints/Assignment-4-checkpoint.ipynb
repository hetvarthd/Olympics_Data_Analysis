```python
import json
import pandas as pd
import matplotlib.pyplot as plt
```

```python
weekly_data = {
    'days': ['Mon', 'Tue', 'Wed', 'Thur', 'Fri', 'Sat', 'Sun'],
    'coding': [2, 1, 2, 3, 2, 4, 4],
    'jogging': [1, 1, 2, 1, 1, 2, 2],
    'studying': [2, 4, 2, 1, 1, 4, 4],
    'workouts': [1, 1, 1, 1, 1, 1, 1],
    'household': [1, 1, 1, 1, 1, 3, 3],
    'lecs': [5, 5, 3, 5, 5, 0, 0]
}
```

```python
with open('Weekly_Activities.json', 'w') as file:
    json.dump(weekly_data, file)
```

```python
df = pd.read_json('Weekly_Activities.json')
df
```

```python
labels = ["Coding", "Jogging", "Studying", "Working out", "Household Chores", "Attending Lectures"]
colors = ['skyblue', 'lightgreen', 'gold', 'lightcoral', 'purple', 'pink']
```

```python
plt.style.use('classic')
plt.figure(figsize=(15,6))
plt.stackplot(df['days'].tolist(), 
              df['coding'].tolist(), 
              df['jogging'].tolist(), 
              df['studying'].tolist(), 
              df['workouts'].tolist(),
              df['household'].tolist(),
              df['lecs'].tolist(),
              labels=labels, colors=colors, alpha=1)
plt.legend(loc='best')
plt.title('Weekly Activity Tracker')
plt.xlabel('Day')
plt.ylabel('Hours')
plt.grid(True)
plt.show()
```
